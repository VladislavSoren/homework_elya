## Тернарный оператор

```python
age = 18

if age >= 18:
    status = "Совершеннолетний"
else:
    status = "Несовершеннолетний"

status = "Совершеннолетний" if age >= 18 else "Несовершеннолетний"  # Если age >= 18, вернет "Совершеннолетний", иначе "Несовершеннолетний"
print(status)
```

```python
# Тестирование функции (концепция тестирования: сравнение ожидаемого результата с фактическим)
def sum_numbers(numbers):
    numbers = [int(number) for number in numbers]
    return sum(numbers)

numbers = [1,"2",3]
axcpect = 6
result = sum_numbers(numbers)
assert axcpect == result
```

```python
import unittest

def sum_numbers(numbers):
    numbers = [int(number) for number in numbers]
    return sum(numbers)

class TestCalk(unittest.TestCase):

    def test_good(self):
        numbers = [1,"2",3]
        self.assertEqual(sum_numbers(numbers), 6)

    def test_fail(self):
        numbers = [1,"2",4]
        self.assertNotEqual(sum_numbers(numbers), 6)


if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
