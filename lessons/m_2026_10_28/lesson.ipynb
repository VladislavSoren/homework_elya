## Тернарный оператор

```python
age = 18

if age >= 18:
    status = "Совершеннолетний"
else:
    status = "Несовершеннолетний"

status = "Совершеннолетний" if age >= 18 else "Несовершеннолетний"  # Если age >= 18, вернет "Совершеннолетний", иначе "Несовершеннолетний"
print(status)
```

```python
l1 = ["Совершеннолетний" if i >= 18 else "Несовершеннолетний" for i in range(16, 20)]
l1
```

```python
l1 = []
for i in range(16, 20):
    if i >= 18:
        l1.append("Совершеннолетний")
    else:
        l1.append("Несовершеннолетний")
l1
```
