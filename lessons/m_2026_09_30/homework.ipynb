## Блок 1. Типы данных и строки

### Теория: типы данных и строки

**Основные типы**

| Тип | Пример | Что хранит |
|---|---|---|
| `int` | `17`, `-3`, `1_000` | целые числа |
| `float` | `1.65`, `-2.5` | дробные числа (неточные!) |
| `str` | `"Эля"`, `'текст'` | строки — последовательность символов |
| `bool` | `True`, `False` | истина / ложь |
| `None` | `None` | «ничего», значение отсутствует |
| `Decimal` | `Decimal("0.1")` | точные дробные числа (деньги) |

**Преобразование типов:** `int("17")` → `17`, `float("1.5")` → `1.5`, `str(10)` → `"10"`. Узнать тип — `type(x)`.

**float неточен:** `0.1 + 0.2` → `0.30000000000000004`. Для денег используют `Decimal` из модуля `decimal`, причём создают его **из строки**: `Decimal("0.1")`.

**Строки**
- индекс — номер символа, начинается с `0`; `-1` — последний символ: `"кот"[0]` → `"к"`, `"кот"[-1]` → `"т"`
- `len(s)` — длина строки
- `s.split()` — делит строку по пробелам и возвращает **список** строк
- `s.lower()` — всё в нижний регистр
- f-строка подставляет значения в текст: `f"Мне {age} лет"`; внутри `{}` можно считать: `f"{age + 1}"`

````python
from decimal import Decimal

# Преобразование типов
a = "3"
b = "2.5"
print(a + a)                # склеивание строк: 33
print(int(a) + float(b))    # сложение чисел: 5.5

# float против Decimal
print(0.1 + 0.1 + 0.1)                                  # 0.30000000000000004
print(Decimal("0.1") + Decimal("0.1") + Decimal("0.1"))  # 0.3

# Индексы и длина строки
word = "Python"
print(word[0], word[-1], len(word))   # P n 6

# split возвращает список строк
parts = "Казань 12 3.5".split()
print(parts)                    # ['Казань', '12', '3.5']
print(int(parts[1]) * 2)        # 24
print(f"Город: {parts[0]}, через год: {int(parts[1]) + 1}")
````

## Блок 2. Условия

### Теория: условия

**Операторы сравнения:** `==` равно, `!=` не равно, `>`, `<`, `>=` (больше или равно), `<=` (меньше или равно). Результат сравнения — `bool`.

**Логические операторы:** `and` (оба условия истинны), `or` (хотя бы одно), `not` (отрицание). Диапазон можно записать так: `10 <= x <= 20`.

**if / elif / else**
```python
if условие1:
    ...        # выполнится, если условие1 True
elif условие2:
    ...        # если условие1 False, а условие2 True
else:
    ...        # если все условия False
```
- условия проверяются **сверху вниз**, выполняется только **первая** подходящая ветка
- поэтому в цепочке `elif` не нужно повторно проверять то, что уже отсечено выше
- блок кода определяется **отступом** (4 пробела)

**Тернарный оператор** — условие в одну строку, возвращает значение:
```python
результат = значение1 if условие else значение2
```
Можно вкладывать: `A if условие1 else (B if условие2 else C)`.

````python
# Цепочка if-elif-else: скорость на дороге
def speed_status(speed):
    if speed < 0:
        return "Ошибка"
    elif speed <= 60:      # сюда попадаем, только если speed >= 0
        return "Норма"
    elif speed <= 80:      # сюда — только если speed > 60
        return "Превышение"
    else:
        return "Штраф"

for s in [-5, 0, 60, 61, 80, 100]:
    print(s, "->", speed_status(s))

# Логические операторы
x = 15
print(10 <= x <= 20)            # True
print(x > 0 and x % 2 == 0)     # False: 15 нечётное

# Тернарный оператор
n = 7
parity = "чётное" if n % 2 == 0 else "нечётное"
print(n, parity)
````

## Блок 3. Циклы и списки

### Теория: циклы и списки

**Список** — упорядоченная изменяемая коллекция: `nums = [3, 1, 2]`. Индексы с `0`, `nums[-1]` — последний элемент.

| Метод | Что делает |
|---|---|
| `append(x)` | добавить в конец |
| `insert(i, x)` | вставить по индексу |
| `remove(x)` | удалить первое вхождение значения |
| `pop()` / `pop(i)` | удалить и вернуть последний / по индексу |
| `sort()` / `sort(reverse=True)` | сортировка на месте (возвращает `None`!) |
| `index(x)`, `count(x)` | индекс первого вхождения, количество вхождений |

**Цикл for** — перебирает элементы коллекции:
- `for x in список:` — по значениям
- `for i in range(start, stop, step):` — по числам, `stop` **не включается**
- `for i, x in enumerate(список):` — сразу индекс и значение (индекс с `0`)

**Цикл while** — повторяет, пока условие `True`. Условие должно когда-то стать `False` (например, счётчик растёт), иначе цикл бесконечный.

**Управление циклом** (работает в `for` и `while`):
- `continue` — пропустить текущий шаг и перейти к следующему
- `break` — полностью выйти из цикла

````python
# for + continue + append
words = ["дом", "солнце", "кот", "машина"]
long_words = []
for w in words:
    if len(w) <= 3:
        continue            # короткие пропускаем
    long_words.append(w)
print(long_words)           # ['солнце', 'машина']

# Методы списков
nums = [4, 1, 3]
nums.sort()
print(nums)                 # [1, 3, 4]

# enumerate: индекс + значение
for i, w in enumerate(["а", "б", "в"]):
    print(i, w)             # 0 а / 1 б / 2 в

# while со счётчиком и break: ищем первое число больше 10
values = [3, 8, 12, 5, 20]
i = 0
while i < len(values):
    if values[i] > 10:
        print("Нашли:", values[i], "на позиции", i)
        break
    i += 1
````

## Блок 4. Словари, множества, кортежи, хэширование

### Теория: словари, множества, кортежи, хэширование

**Словарь** `dict` — пары «ключ: значение»: `d = {"кот": 3}`
- чтение: `d["кот"]`; добавление/изменение: `d["пёс"] = 1`; удаление: `del d["кот"]`
- проверка ключа: `"кот" in d`
- перебор: `for key, value in d.items():`

**Кортеж** `tuple` — упорядоченная **неизменяемая** коллекция: `t = (1, 2)`
- распаковка: `a, b = (1, 2)`; в цикле: `for a, b in [(1, 2), (3, 4)]:`
- функция может вернуть несколько значений — они вернутся кортежем: `return x, y`

**zip** — идёт по нескольким спискам одновременно, собирая элементы в кортежи:
`zip(["a", "b"], [1, 2])` → `("a", 1), ("b", 2)`

**Множество** `set` — неупорядоченная коллекция **уникальных** элементов, повторы удаляются: `set([1, 1, 2])` → `{1, 2}`

- `a & b` — пересечение (есть в обоих)
- `a | b` — объединение (есть хотя бы в одном)
- `a - b` — разность (есть в `a`, но нет в `b`)
- `s.add(x)` — добавить элемент

**Хэширование:** ключи словаря и элементы множества должны быть **хэшируемыми**, то есть неизменяемыми: числа, строки, кортежи, `bool`, `None`, `Decimal`. Изменяемые объекты (список, словарь, множество) хэшировать нельзя — `hash([1, 2])` даёт `TypeError`.

````python
# Словарь: добавление, изменение, перебор
ages = {"Аня": 20}
ages["Борис"] = 25          # добавили
ages["Аня"] += 1            # изменили
for name, age in ages.items():
    print(name, age)

# zip + распаковка кортежей
cities = ["Казань", "Омск"]
people = [1300, 1100]
for city, count in zip(cities, people):
    print(city, count)

# Функция возвращает кортеж -> распаковка
def min_max(values):
    return min(values), max(values)

low, high = min_max([5, 2, 9])
print(low, high)            # 2 9

# Множества
a = {1, 2, 3, 3}            # повтор исчезнет
b = {3, 4}
print(a, a & b, a | b, a - b)

# Хэшируемость
print(hash((1, 2)))         # кортеж можно
# hash([1, 2])              # список нельзя -> TypeError
````

## Блок 5. Функции, аргументы, LEGB

### Теория: функции, аргументы, LEGB

**Функция** — именованный блок кода, который можно вызывать много раз:
```python
def имя(параметр1, параметр2="по умолчанию"):
    ...
    return результат
```
- `return` **сразу завершает** функцию и возвращает значение; код после него не выполняется
- если `return` нет — функция возвращает `None`
- `print` только показывает значение на экране, а `return` отдаёт его наружу, чтобы сохранить в переменную

**Аргументы**
- **позиционные** — сопоставляются по порядку: `f(1, 2)`
- **именованные** — по имени параметра: `f(a=1, b=2)`
- **параметр по умолчанию** — используется, если аргумент не передали: `def f(a, b=10)`; параметры по умолчанию пишутся **после** обычных

**LEGB** — порядок, в котором Python ищет имя переменной:
1. **L**ocal — внутри текущей функции (включая её параметры)
2. **E**nclosed — во внешней функции (если функция вложенная)
3. **G**lobal — на уровне файла/ноутбука
4. **B**uilt-in — встроенные имена: `print`, `len`, `int`...

Поиск останавливается на **первом** найденном имени. Присваивание внутри функции создаёт **локальную** переменную и не меняет глобальную.

````python
# Параметр по умолчанию, позиционные и именованные аргументы
def power(base, exp=2):
    return base ** exp

print(power(3))             # 9   — exp по умолчанию
print(power(3, 3))          # 27  — позиционно
print(power(exp=4, base=2)) # 16  — по имени, порядок не важен

# return завершает функцию
def first_negative(values):
    for v in values:
        if v < 0:
            return v        # дальше цикл не идёт
    return None             # если отрицательных нет

print(first_negative([3, -1, -5]))  # -1
print(first_negative([1, 2]))       # None

# LEGB
color = "красный"           # Global

def paint():
    color = "синий"         # Local для paint, Enclosed для brush

    def brush():
        print("brush:", color)   # L нет -> E: синий

    brush()

paint()
print("global:", color)     # красный — глобальная не изменилась
print(len("LEGB"))          # len — Built-in
````
