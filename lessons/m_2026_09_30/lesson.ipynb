```python
cities = ["Казань", "Омск"]
people = [1300, 1100]
for city, count in zip(cities, people):
    print(city, count)
```

```python
print(list(zip(cities, people)))

"""
zip внутри себя формирует таблицу вида (метафора)
"Казань", 1300
"Омск", 1100

а по факту zip создаёт список кортежей вида [('Казань', 1300), ('Омск', 1100)]
"""
```

```python
cities = ["Казань", "Омск"]
people = [1300, 1100]
ages = [1000, 500]
for city, count, age in zip(cities, people, ages):
    print(city, count, age)
```

```python
cities = ["Казань", "Омск", "Орёл"]
people = [1300, 1100]
for city, count in zip(cities, people):
    print(city, count)
```

```python
cities = ["Казань", "Омск", "Орёл"]
people = [1300, 1100]
for city, count in zip(cities, people, strict=True):
    print(city, count)
```
