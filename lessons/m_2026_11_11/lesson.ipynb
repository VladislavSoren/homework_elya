## Декораторы с аргументами

```python
from time import time

def timer(fun):
    def wrapper(*args, **kwargs):
        start_time = time()
        fun(*args, **kwargs)
        exec_time = time() - start_time
        print("Время выполнения: ", exec_time)
    return wrapper
```

```python
from time import sleep

def fun(word_start, word_end):
    print(word_start)
    sleep(2)
    print(word_end)

fun("Начало", "Конец")
```

```python
def timer(porog):
    def wrapper(fun):
        def inner_wrapper(*args, **kwargs): # *args -  означает "распакованные" позиционные аргументы 1,2,3
            # args  # Но если обращаться без *, то уни буду упаковываться в кортеж (1,2,3)
            start_time = time()
            print(args[0])
            print(args[1])
            fun(*args, **kwargs)
            exec_time = time() - start_time
            print("Время выполнения: ", exec_time)

            if exec_time > porog: 
                print("Время выполнения слишком велико")
        return inner_wrapper
    return wrapper
```

```python
def fun(word_start, word_end):
    print(word_start)
    sleep(2)
    print(word_end)

wrapper = timer(1)
inner_wrapper = wrapper(fun)
inner_wrapper("Начало", "Конец")
```

```python
@timer(porog=1)
def fun(word_start, word_end):

    print(word_start, word_end)
    
    print(word_start)
    sleep(2)
    print(word_end)

fun("Начало", "Конец")
```

```python
@timer(porog=1)
@timer(porog=3)
def fun(word_start, word_end):

    print(word_start, word_end)
    
    print(word_start)
    sleep(1)
    print(word_end)

fun("Начало", "Конец")
```
