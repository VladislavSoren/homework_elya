```python
"""
### Задача 2. Функция + цикл + список

Написать функцию count_long_words, которая:
- принимает список слов
- возвращает кол-во слов, длина которых больше 5 букв

Пример:
count_long_words(["кот", "черепаха", "дом", "программист"])  ->  2
"""
```

```python
def count_long_words(words):
    count = 0
    for word in words:
        if len(word) > 5:
            count += 1
    return count
print(count_long_words(["кот", "черепаха", "дом", "программист"]))
```

```python
def count_long_words(words, porog):
    count = 0
    for word in words:
        if len(word) > porog:
            count += 1
    return count
print(count_long_words(["кот", "черепаха", "дом", "программист"], porog=1))
```

```python
"""
### Задача 3. Аргумент по умолчанию

Написать функцию greet, которая:
- принимает имя и приветствие (по умолчанию "Привет")
- возвращает строку вида "Привет, Эля!"

Пример:
greet("Эля")               ->  "Привет, Эля!"
greet("Эля", "Здравствуй") ->  "Здравствуй, Эля!"
"""
```

```python
def greet(name):
    return "Привет, " + name + "!"
greet("Эля")
```

```python
def greet(name, greeting):
    return greeting + ", "  + name + "!"

greet("Эля", "Здравствуй")
```

```python
"""
### Задача 4. Функция + словарь

Есть список покупок:
purchases = [
    {"name": "хлеб", "price": 50, "count": 2},
    {"name": "молоко", "price": 90, "count": 1},
    {"name": "сыр", "price": 300, "count": 3},
]

Написать функцию total_sum, которая:
- принимает такой список
- возвращает общую стоимость всех покупок (price * count по каждой)
"""
```

```python
purchases = [
    {"name": "хлеб", "price": 50, "count": 2},
    {"name": "молоко", "price": 90, "count": 1},
    {"name": "сыр", "price": 300, "count": 3},
]

def total_sum(purchases):
    total = 0
    for purchase in purchases:
        total += purchase["price"] * purchase["count"]
    return total

print(total_sum(purchases))
```

```python
"""
### Задача под звёздочкой

Написать функцию word_stat, которая:
- принимает текст
- возвращает словарь, где ключ — слово (в нижнем регистре), значение — сколько раз оно встретилось

Пример:
word_stat("Кот кот собака")  ->  {"кот": 2, "собака": 1}
"""
```

```python
def word_stat(text):
    result = {}

    words = text.split()

    for word in words:
        word = word.lower()

        if word not in result:
            result[word] = 1
        else:
            result[word] += 1
        return result
    
print(word_stat("Кот кот собака"))
```
