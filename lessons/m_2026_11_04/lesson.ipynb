## Декораторы

```python
def decorator(func):
    def wrapper():
        print("Начало")
        func()
        print("Конец")
    return wrapper
```

```python
def func():
    print("Работаю")

func()
```

```python
@decorator
def func():
    print("Работаю")

func()
```

```python
def func():
    print("Работаю")

# Как происходит декорирование
func = decorator(func)
func()
```

```python
from time import time

def timer(fun):
    def wrapper():
        start_time = time()
        fun()
        exec_time = time() - start_time
        print("Время выполнения: ", exec_time)
    return wrapper
```

```python
from time import sleep

def fun():
    print("Начало")
    sleep(2)
    print("Конец")

fun()
```

```python
@timer
def fun():
    print("Начало")
    sleep(2)
    print("Конец")

fun()
```
