```python
def count_words(text: str):
    words = text.split(sep=" ")
    count = 0
    for word in words:
        count = count + 1
    return f"количество слов: {count}"

count_words("Маша едет в лагерь отдыхать")
```

```python
# посчитать гласные буквы
def count_vowels(text: str):
    vowels = text.split(sep=" ")
    count = 0
    for vowel in vowels:
        count = count + 1
    return("количество гласных: ")

count_vowels("тфвофдывшфодлю")
```
