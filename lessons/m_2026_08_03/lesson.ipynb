```python
def count_sentences(text: str):
    count = text.count(". ") + text.count("!") + text.count("?") 
    #count = text.count(". ") + text.count("!") + text.count("?") + text.count("...") + text.count(".")
    if text[-1] == ".":
        count += 1
    print(f"В тексте {count} предложений")

    if count > 5:
       print("Длинный текст")

    if text.count("!") > 2:
        print("Текст воодушевляющий")

count_sentences("Привет! Меня зовут Павел! Я пошел гулять... Как дела? Хочу есть. Почему молчишь? Пока.")
```
